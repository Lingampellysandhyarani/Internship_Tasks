```python
# Q35. Prediction: [[1, 99], [3, 4]]

import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Explanation: copy.copy() makes a shallow copy, so both lists share the inner lists.
```

```python
# Q36. Prediction: [[1, 2], [3, 4]]

import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# Explanation: deepcopy() creates separate inner lists, so changing b does not affect a.
```
