## Q1. Find the Shortest Word in a Sentence

### 1. Observe
Input: Python makes coding easy  
Output: easy

### 2. Understand
Find the word with the smallest length in the sentence.

### 3.Break

### 4. Logic (Pseudocode)
1. Start
2. Take the sentence
3. Split the sentence into words
4. Set the first word as the shortest word
5. Take each word
6. Compare its length with the shortest word
7. If it is shorter, update the shortest word
8. Print the shortest word
9. Stop

### 5.Code

```python
#Step 5
sentence = "Python makes coding easy"
words = sentence.split()

shortest = words[0]

for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)
```

## Q8. Find the Most Frequent Character

### 1. Observe
Input: banana

Output: a

### 2. Understand
Find the character that appears most often. Ignore spaces.



### 3.Break

### 4. Logic (Pseudocode)
1. Start
2. Take the string
3. Count each character
4. Find the highest frequency
5. Select the first character with that frequency
6. Print the character
7. Stop

### 5.code

```python
def most_frequent(text):
    text = text.replace(" ", "")
    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    return max(frequency, key=frequency.get)

print(most_frequent("banana"))
```

### 6.Test

```python
text = "hello"
text = text.replace(" ", "")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print(max(frequency, key=frequency.get))
```

```python
def repeated_chars(text):
    text = text.replace(" ", "")
    repeated = []

    for char in text:
        if text.count(char) > 1 and char not in repeated:
            repeated.append(char)

    return repeated

print(repeated_chars("programming"))
```

### 1. Observe
Input: [1, 2, 3, 4, 5]  
Output: {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}

### 2. Understand
Create a dictionary with each number as a key and its status (Even or Odd) as the value.


### 3. Break


### 4. Logic (Pseudocode)
1. Start
2. Take a list of numbers
3. Create an empty dictionary
4. Take each number
5. Check if the number is divisible by 2
6. If yes, store 'Even'; otherwise, store 'Odd'
7. Print the dictionary
8. Stop

### 5.Code

```python
def even_odd_status(numbers):
    result = {}
    for num in numbers:
        if num % 2 == 0:
            result[num] = "Even"
        else:
            result[num] = "Odd"
    return result

print(even_odd_status([1, 2, 3, 4, 5]))
```

## Q17.  Find All Words Appearing More Than Once

### 1. Observe
Input: data science uses data and science tools  
Output: ['data', 'science']

### 2. Understand
Find repeated words, ignoring uppercase and lowercase differences. Display each repeated word only once, in order of first appearance.


### 3. Break

### 4. Logic (Pseudocode)
1. Start
2. Take the sentence
3. Convert it to lowercase and split into words
4. Find words that appear more than once
5. Store each repeated word only once
6. Print the result
7. Stop

### 5.Code

```python
def repeated_words(sentence):
    words = sentence.lower().split()
    result = []

    for word in words:
        count = words.count(word)

        if count > 1:
            if word not in result:
                result.append(word)

    return result

print(repeated_words("data science uses data and science tools"))
```

### 6.Test

## Q18. Find the First Word That Appears Only Once

### 1. Observe
Input: apple banana apple mango banana orange  
Output: mango

### 2. Understand
Find the first word that appears only once, ignoring uppercase and lowercase differences.


### 3.Break

### 4. Logic (Pseudocode)
1. Start
2. Take the sentence
3. Convert it to lowercase
4. Split the sentence into words
5. Take each word
6. Count how many times it appears
7. If the count is 1, return the word
8. Stop

### 5.Code

```python
def first_unique_word(sentence):
    words = sentence.lower().split()

    for word in words:
        count = words.count(word)

        if count == 1:
            return word

    return None

print(first_unique_word("apple banana apple mango banana orange"))
```

## Q20. Find the Longest Word That Appears Only Once

### 1. Observe
Input: Python makes coding useful Python  
Output: coding

### 2. Understand
Find the longest word that appears only once. If multiple words have the same length, choose the first one.


### 3.Break

### 4. Logic (Pseudocode)
1. Start
2. Take the sentence
3. Convert it to lowercase and split into words
4. Create an empty variable for the longest word
5. Take each word
6. Check if it appears only once
7. If it is longer than the current longest word, update it
8. Print the longest word
9. Stop

### 5. Code

```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    longest = ""

    for word in words:
        if words.count(word) == 1:
            if len(word) > len(longest):
                longest = word

    return longest

print(longest_unique_word("Python makes coding useful Python"))
```
