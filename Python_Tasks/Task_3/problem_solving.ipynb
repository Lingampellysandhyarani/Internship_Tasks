#### Q1. Find the Frequency of Each Word

**Pseudocode:**

1. Take a sentence.
2. Split the sentence into words.
3. Create an empty dictionary.
4. Check each word one by one.
5. If the word is already there, increase its count by 1.
6. If it is not there, add the word with count 1.
7. Return the dictionary.

```python
def word_frequency(sentence):
    words = sentence.split()
    counts = {}

    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

    return counts


#Test case
sentence = "python is easy and python is powerful"
print(word_frequency(sentence))

# Additional test case
sentence2 = "data is important and data is interesting"
print(word_frequency(sentence2))
```

#### Q2. Find the Frequency of Each Character


**Pseudocode:**

1. Take a string.
2. Make an empty dictionary.
3. Look at each character one by one.
4. If the character is already in the dictionary, add 1 to its count.
5. If it is new, add it to the dictionary with count 1.
6. Return the dictionary with each character's count.

```python
def character_frequency(string):
    counts = {}

    for char in string:
        if char in counts:
            counts[char] += 1
        else:
            counts[char] = 1

    return counts


#Test case
string = "banana"
print(character_frequency(string))

# Additional test case
string2 = "data analysis"
print(character_frequency(string2))
```

#### Q3. Find the First Non-Repeating Character


**Pseudocode:**

1. Take a string.
2. Check each character from left to right.
3. Count how many times the character appears.
4. If it appears only once, return that character.
5. If every character repeats, return None.

```python
def first_non_repeating(string):
    for char in string:
        if string.count(char) == 1:
            return char

    return None


# Original test case
string = "swiss"
print(first_non_repeating(string))

# Additional test case
string2 = "aabbcc"
print(first_non_repeating(string2))
```

#### Q4. Check Whether Two Strings Are Anagrams


**Pseudocode:**

1. Take two strings.
2. Arrange the characters of both strings in the same order.
3. Compare the two strings.
4. If they are the same, return True.
5. Otherwise, return False.

```python
def are_anagrams(s1, s2):
    if sorted(s1) == sorted(s2):
        return True
    else:
        return False


# Original test case
print(are_anagrams("listen", "silent"))

# Additional test case
print(are_anagrams("hello", "world"))
```

#### Q5. Find the First Repeating Character


**Pseudocode:**

1. Take a string.
2. Create an empty set to remember the characters.
3. Check each character from left to right.
4. If the character is already in the set, return it.
5. Otherwise, add the character to the set.
6. If no character repeats, return None.

```python
def first_repeating(Text):
  characters = set()

  for char in Text:
    if char in characters:
      return char
    characters.add(char)

  return None

# Test case

print(first_repeating("programming"))

# Additional Test case

print(first_repeating("apple"))
```

#### Q6. Move All Zeros to the End of a List


**Pseudocode:**

1. Take a list of numbers.
2. Make a new empty list.
3. Add all non-zero numbers to the new list.
4. Count the zeros in the original list.
5. Put all the zeros at the end.
6. Return the new list.

```python
def move_zeros(nums):

    result = []

    for num in nums:
          if num != 0:
             result.append(num)
    
    zero_count =len(nums) - len(result)
    result.extend([0]* zero_count)
            
    return result
    
# Test case
nums = [0, 3, 0, 1, 2, 0, 4]

print(move_zeros(nums))

# Additional test case

nums1=[0,0,5,0,2,0,1]

print(move_zeros(nums1))
```

#### Q7. Find the Second Largest Number in a List


**Pseudocode:**

1. Take a list of numbers.
2. Remove duplicate numbers.
3. Sort the numbers in descending order.
4. If there are at least two unique numbers, take the second number.
5. Return the result.

```python

def second_highest(numbers):
    unique_number = sorted(set(numbers), reverse=True)

    if len(unique_number) < 2:
        return None

    return unique_number[1]


# Original test case
numbers = [10, 20, 40, 30, 40, 20]
print(second_highest(numbers))

# Additional test case
nums = [12, 18, 22, 89, 40, 12, 13]
print(second_highest(nums))


```

#### Q8. Remove Duplicates Without Changing the Original Order


**Pseudocode:**

1. Take a list of numbers.
2. Create an empty list.
3. Check each number in the original list.
4. If the number is not already in the new list, add it.
5. Return the new list.



```python
def remove_duplicate(list):
    result = []

    for num in list:
        if num not in result :
            result.append(num)

    return result

# Test case

list = [4,2,4,5,2,7,5]

print(remove_duplicate(list))

# Additional Test case

list1 = [ 1,4,6,3,2,1,2,9,3]

print(remove_duplicate(list1))
```

#### Q9. Find the Missing Number from 1 to N


**Pseudocode:**

1. Take a list of numbers.
2. Find the total number that should be present in the list.
3. Calculate the expected sum of numbers from 1 to n.
4. Calculate the sum of the numbers in the given list.
5. Subtract the actual sum from the expected sum.
6. Return the missing number.

```python
def find_missing(numbers):
    n = len(numbers) + 1
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(numbers)

    return expected_sum - actual_sum


#Test case
print(find_missing([1, 2, 4, 5, 6]))

# Additional test case
print(find_missing([1, 3,4, 5]))
```

#### Q10. Find Two Numbers Whose Sum Equals a Target


**Pseudocode:**

1. Take a list of numbers and a target value.
2. Check each number from left to right.
3. For each number, check the numbers that come before it.
4. Add the current number and each earlier number.
5. If their sum equals the target, return the pair.
6. If no pair matches the target, return an empty list.

```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i):
            if nums[j] + nums[i] == target:
                return [nums[j], nums[i]]

    return []


# Test case
print(two_sum([2, 7, 11, 15], 9))

# Additional test case
print(two_sum([1, 4, 6, 8], 10))

# No valid pair
print(two_sum([1, 2, 3], 10))
```

#### Q11. Find the Number Closest to Zero


**Pseudocode:**

1. Take a list of numbers.
2. Assume the first number is the closest to zero.
3. Check each number in the list.
4. Compare the absolute values of the numbers.
5. If a number is closer to zero, update the closest number.
6. If two numbers are equally c to zero, choose the positive number.
7. Return the closest number.

```python
def closest_to_zero(numbers):
    closest = numbers[0]

    for num in numbers:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num

    return closest


# Original test case
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Additional test case
print(closest_to_zero([-5, -1, 4, 2]))

# Another test case
print(closest_to_zero([-7, -3, 3, 8]))
```

#### Q12. Find Common Elements Between Two Lists


**Pseudocode:**

1. Take two lists.
2. Create an empty result list.
3. Check each element in the first list.
4. Check whether the element is present in the second list.
5. If it is present and not already in the result list, add it.
6. Return the result list.

```python
def common_elements(list1, list2):
    result = []

    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result


# Test case
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Additional test case
print(common_elements([4, 2, 4, 1, 3], [4, 3, 2]))
```

#### Q13. Rotate a List to the Right by K Positions


**Pseudocode:**

1. Take a list of numbers and a value K.
2. If the list is empty, return the list.
3. Calculate the effective number of rotations using K modulo the list length.
4. Take the last K elements of the list.
5. Move those elements to the beginning.
6. Add the remaining elements after them.
7. Return the rotated list.

```python
def rotate_right(nums, k):
    if len(nums) == 0:
        return nums

    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k != 0 else nums


#Test case
print(rotate_right([1, 2, 3, 4, 5], 2))

# Additional test case
print(rotate_right([10, 20, 30, 40], 1))
```

#### Q14. Find the Maximum Sum of Two Consecutive Elements


**Pseudocode:**

1. Take a list containing at least two numbers.
2. Add the first two numbers and store their sum as the maximum.
3. Check each pair of consecutive numbers in the list.
4. Add each pair and compare its sum with the current maximum.
5. If the new sum is greater, update the maximum.
6. Return the maximum sum.

```python
def max_consecutive_sum(numbers):
    maximum = numbers[0] + numbers[1]

    for i in range(1, len(numbers) - 1):
        current_sum = numbers[i] + numbers[i + 1]

        if current_sum > maximum:
            maximum = current_sum

    return maximum


#Test case
print(max_consecutive_sum([4, 7, 2, 9, 5]))

# Additional test case
print(max_consecutive_sum([1, 8, 3, 6]))
```

#### Q15. Find All Pairs of Numbers Whose Sum Equals a Target


**Pseudocode:**

1. Take a list of numbers and a target value.
2. Create an empty list to store the pairs.
3. Check each number with the numbers that come after it.
4. If their sum equals the target, arrange the pair in ascending order.
5. Add the pair if it is not already in the result list.
6. Sort the result pairs by their first element.
7. Return the result list.

```python
def find_all_pairs(nums, target):
    result = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])

                if pair not in result:
                    result.append(pair)

    result.sort(key=lambda pair: pair[0])
    return result


# Test case
print(find_all_pairs([2, 4, 3, 5, 7, 8, 1], 9))

# Additional test case
print(find_all_pairs([1, 2, 3, 4, 5], 6))
```
